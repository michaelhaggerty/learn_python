# Python Lists


What is a list?

- one of 4 built-in data types in Python used to store collections of data, the other 3 are Tuple, Set, and Dictionary, all with different qualities and usage.
- an ordered, mutable, and heterogeneous collection used to store multiple items in a single variable. Lists are defined using square brackets [] with elements separated by commas.
- At the architectural level, a Python list is not a traditional linked list and it does not store values directly inside a contiguous block of memory. Instead, it is a dynamic array of object references (pointers).
  List items are ordered, changeable, and allow duplicate values.
- List items are indexed, the first item has index [0], the second item has index [1] etc.

Are there list methods that change the order?

- Note: There are some list methods that will change the order, but in general: the order of the items will not change.


How to create a list?

- Lists can be created by placing a comma-separated sequence of values inside square brackets [].


```python
test_list = ['a', 'b', 'c']
print(test_list)
```

Can lists have items with same value?

- yes


```python
thislist = ["apple", "banana", "cherry", "apple", "cherry"]
print(thislist)
```

How to determine the number of items a list has?

- use len() function to determine the number of items a list has.


```python
len(test_list)
```

What data type can list items be?

- List items can be of any data type, including numbers, strings, and even other lists


```python
list1 = ["a", "b", "v"]
list2 = [1, 2, 2]
list3 = [True, False, False]
list4 = ["abc", 34, True, 40, "male"]
list5 = [["a", "b", "v"], [1, 2, 2]]
```

What data type are lists and how can you determine an object's data type?

- Lists are a data type of their own, and you can determine an object's data type using the type() function. For example, type(my_list) will return <class 'list'> if my_list is a list.


```python
type(test_list)
```

What is the list() constructor?

- The list() constructor is a built-in function in Python that allows you to create a new list object. It can be used to convert other iterable data types (like tuples, sets, or strings) into a list.


```python
# note the double round-brackets
thislist = list(("apple", "banana", "cherry"))
print(thislist)
```

What are collections in Python?

- collections refers to two closely related concepts: the general category of container data structures that group multiple objects together, and the built-in collections module, which provides specialized, high-performance alternatives to Python’s four standard containers.


What are the four standard containers in Python?
There are four collection data types in the Python programming language:

1. List is a collection which is ordered and changeable. Allows duplicate members.
2. Tuple is a collection which is ordered and unchangeable. Allows duplicate members.
3. Set is a collection which is unordered, unchangeable\*, and unindexed. No duplicate members.
4. Dictionary is a collection which is ordered\*\* and changeable. No duplicate members.


| Collection Type | Ordered | Changeable (Mutable)   | Indexed          | Duplicate Members  |
| :-------------- | :------ | :--------------------- | :--------------- | :----------------- |
| **List**        | Yes     | Yes                    | Yes (by integer) | Allowed            |
| **Tuple**       | Yes     | No                     | Yes (by integer) | Allowed            |
| **Set**         | No      | No\* (Add/remove only) | No               | Not allowed        |
| **Dictionary**  | Yes\*\* | Yes                    | Yes (by key)     | Not allowed (keys) |

_\*Set items are unchangeable, but you can remove and/or add items whenever you like._  
_\*\*As of Python version 3.7, dictionaries are ordered. In Python 3.6 and earlier, dictionaries are unordered._


# Access List Items


How to access list items?

- List items are indexed and you can access them by referring to the index number:

What index does the first item have?

- The first item has index 0.


```python
thislist = ["apple", "banana", "cherry"]
print(thislist[1])
```

What is negative indexing?

- Negative indexing means beginning from the end, -1 refers to the last item, -2 refers to the second last item etc.


```python
# print the last item of the list

thislist = ["apple", "banana", "cherry"]
print(thislist[-1])
```

How to specify a range of indexes?

- You can specify a range of indexes by specifying where to start and where to end the range


```python
thislist = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
print(thislist[2:5])
```

Where does the previous search start and end?

- The search will start at index 2 (included) and end at index 5 (not included).

What happens if we leave out the start index?

- The search will start at the first item in the list.


```python
# This example returns the items from the beginning to, but NOT including, "kiwi":
thislist = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
print(thislist[:4])
```

What happens if we leave out the end index?

- By leaving out the end value, the range will go on to the end of the list


```python
# This example returns the items from "cherry" to the end
thislist = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
print(thislist[2:])
```

What if you want to start the search form the end of the list?

- Specify negative indexes if you want to start the search from the end of the list


```python
# This example returns the items from "orange" (-4) to, but NOT including "mango" (-1):

thislist = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
print(thislist[-4:-1])
```

How to check if item is in a list?

- To determine if a specified item is present in a list use the in keyword


```python
# Check if "apple" is present in the list
thislist = ["apple", "banana", "cherry"]
if "apple" in thislist:
    print("Yes, 'apple' is in the fruits list")
```

```python
fruits = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
print(fruits[0:4])
```

# Change list items


How to change the value of a specific item in a list?

- Specify the index of the item you want to change and assign it a new value


```python
# change the 2nd item
thislist = ["apple", "banana", "cherry"]
thislist[1] = "blackcurrant"
print(thislist)
```

How to change a range of item values?

- define a list with the new values, and refer to the range of index numbers where you want to insert the new value


```python
# Change the values "b" and "b" with the values "1" and "2":

thislist = ["a", "b", "c", "d", "e"]
thislist[1:3] = ["new1", "new2"]
print(thislist)
```

What happens if you insert more items?

- If you insert more items than you replace, the new items will be inserted where you specified, and the remaining items will move accordingly

What happens to the length of the list?

- The length of the list will change when the number of items inserted does not match the number of items replaced.


```python
# Change the second value by replacing it with two new values:
thislist = ["apple", "banana", "cherry"]
thislist[1:2] = ["new1", "new2"]
print(thislist)
```

What happens if you insert less?

- If you insert less items than you replace, the new items will be inserted where you specified, and the remaining items will move accordingly


```python
thislist = ["apple", "banana", "cherry"]
thislist[1:3] = ["watermelon"]
print(thislist)
```

How to insert a new list item without replacing any of the existing values?

- The insert() method inserts an item at the specified index


```python
test_list = ['a', 'b', 'c']
test_list.insert(2, 1)
print(test_list)
print(len(test_list))
```

```python
# Insert "watermelon" as the third item:
thislist = ["apple", "banana", "cherry"]
thislist.insert(2, "watermelon")
print(thislist)
```

# Add list items


How to add item to the end of the list?

- use the append() method


```python
thislist = ["apple", "banana", "cherry"]
thislist.append("orange")
print(thislist)
```

How to insert a list item at a specified index?

- use the insert() method.


```python
# Insert an item as the second position
# Note: As a result of the examples above, the lists will now contain 4 items.

thislist = ["apple", "banana", "cherry"]
thislist.insert(1, "orange")
print(thislist)
```

How to append elements from another list to the current list? Where items added?

- use the extend() method.
- The elements will be added to the end of the list.


```python
# Add the elements of tropical to thislist:


thislist = ["apple", "banana", "cherry"]
tropical = ["mango", "pineapple", "papaya"]
thislist.extend(tropical)
print(thislist)
```

How to Add Any Iterable?

- The extend() method does not have to append lists, you can add any iterable object (tuples, sets, dictionaries etc.).


```python
# Add elements of a tuple to a list:

thislist = ["apple", "banana", "cherry"]
thistuple = ("kiwi", "orange")
thislist.extend(thistuple)
print(thislist)
```

# Remove list items


How to remove a specified item from a list?

- use the remove() method. This method removes the first occurrence of the specified value.


```python
# Remove "banana":

thislist = ["apple", "banana", "cherry"]
thislist.remove("banana")
print(thislist)
```

What happens if there are more than one item with the specified value?

- If there are more than one item with the specified value, the remove() method removes the first occurrence:


```python
# Remove the first occurrence of "banana":

thislist = ["apple", "banana", "cherry", "banana", "kiwi"]
thislist.remove("banana")
print(thislist)
```

How to remove by index?

- The pop() method removes the specified index.


```python
# Remove the second item:

thislist = ["apple", "banana", "cherry"]
thislist.pop(1)
print(thislist)
```

What happens if you don't specify?

- If you do not specify the index, the pop() method removes the last item.


```python
# Remove the last item:

thislist = ["apple", "banana", "cherry"]
thislist.pop()
print(thislist)
```

What about the del keyword?

- The del keyword removes the specified index, or can delete the list completely.


```python
# Remove the first item:

thislist = ["apple", "banana", "cherry"]
del thislist[0]
print(thislist)
```

What else can the del keyword do?

- The del keyword can also delete the list completely.


```python
# Delete the entire list

thislist = ["apple", "banana", "cherry"]
del thislist
print(thislist)
```

How to clear the list? Difference from del()?

- The clear() method empties the list.
- The list still remains, but it has no content.


```python
# Clear the list content:

thislist = ["apple", "banana", "cherry"]
thislist.clear()
print(thislist)
```

```python
fruits = [1, 2, 3]
print(fruits)
fruits.clear()
print(fruits)
```

# Loop Lists


How to loop through a list?

- You can loop through a list using a for loop, which allows you to iterate over each item in the list and perform actions on them. For example:


```python
# Print all items in the list, one by one:

thislist = ["apple", "banana", "cherry"]
for x in thislist:
    print(x)
```

How to loop through index numbers?

- You can also loop through the list items by referring to their index number.
- Use the range() and len() functions to create a suitable iterable.


```python
# Print all items by referring to their index number:

thislist = ["apple", "banana", "cherry"]
for i in range(len(thislist)):
    print(thislist[i])
```

Can you loop through a list using a while loop?

- You can loop through the list items by using a while loop.

- Use the len() function to determine the length of the list, then start at 0 and loop your way through the list items by referring to their indexes.

- Remember to increase the index by 1 after each iteration.


```python
list1 = [1, 2, 3]
i = 0

while i < len(list1):
    print(list1[i])
    i += 1
```

What about list comprehension?

- a concise, single-line syntax for creating a new list by transforming or filtering elements from an existing iterable (like a list, range, string, or tuple). It draws directly from mathematical set-builder notation (e.g., $\{x^2 \mid x \in \mathbb{N}, x \text{ is even}\}$).
- List Comprehension offers the shortest syntax for looping through lists:


```python
# A short hand for loop that will print all items in a list:

thislist = ["apple", "banana", "cherry"]
[print(x) for x in thislist]
```
