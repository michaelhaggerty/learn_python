# Variables


What are variables?

- containers for storing data values

How to create a variable in Python?

- you just use the assignment operator `=` to assign a value to a variable name.
- it must be "valid" python variable name (no spaces, cannot start with a number, cannot use reserved keywords, etc.)


`````python
# examples
x = 5
y = "John"
print(x)
print(y)
`````

What are types in programming? Do they need to be declared in Python?

- types are categories of data that determine what kind of values a variable can hold and what operations can be performed on them.
- in Python, you do not need to declare the type explicitly.
- - Python is "dynamically typed", which means that the type of a variable is inferred based on the value assigned to it.
- - However, you can use type hints to indicate the expected type of a variable, but it is not mandatory.


`````python
x = 4
print(x)
x = "Sally"
print(x)
`````

What is casting?

- the process of converting a variable from one type to another. In Python, you can use built-in functions like `int()`, `float()`, and `str()` to cast variables to different types.
- For example, you can convert a string to an integer using `int("10")`, or convert a float to a string using `str(3.14)`.


`````python
x = str(3)    # x will be '3'
y = int(3)    # y will be 3
z = float(3)  # z will be 3.0
`````

How can you find the type of a variable in Python?

# Variable Names


What are the rules for naming variables in Python?

1. Variable names must start with a letter (a-z, A-Z) or an underscore (\_). CANNOT start with a number (0-9).
2. A variable name can only contain letters, numbers, and underscores (a-z, A-Z, 0-9, and _). A variable name can only contain alpha-numeric characters and underscores (A-z, 0-9, and _ )
3. Variable names are case-sensitive (age, Age and AGE are three different variables)
4. A variable name cannot be any of the Python keywords.


`````python
# examples
myvar = "John"
my_var = "John"
_my_var = "John"
myVar = "John"
MYVAR = "John"
myvar2 = "John"
`````

`````python
# illegal names
2myvar = "John"
my-var = "John"
my var = "John"
`````

# Multi-word variable names


What is best practice for multi-word variable names in Python?

1. camel case: first word is lowercase, subsequent words are capitalized (e.g., myVariableName)
2. snake case: all words are lowercase and separated by underscores (e.g., my_variable_name)
3. pascal case: all words are capitalized (e.g., MyVariableName)


`````python
# camel case
myVariableName = "text"

# snake case
my_variable_name = "text"

# pascal case
MyVariableName = "text"
`````

### Why Python Adds a Space in `print()`

By default, Python's built-in `print()` function includes a parameter called `sep` (short for separator).

When you pass multiple arguments separated by commas, the `sep` parameter determines what string is inserted between them. Its default value is a single space (`' '`).

#### How to Override the Default Space

You can easily change this behavior by explicitly defining the `sep` parameter in your print statement.

````python
# Default behavior (sep=' ' is implied)
print('Hello', 'World')
# Output: Hello World

# 1. Remove the space completely
print('Hello', 'World', sep='')
# Output: HelloWorld

# 2. Use a custom character (e.g., a dash)
print('Hello', 'World', sep='-')
# Output: Hello-World

# 3. Use a newline character to print on separate lines
print('Hello', 'World', sep='\n')
# Output:
# Hello
# World
# 4. if you add two strings together using the `+` operator, Python will concatenate them without adding any spaces. For example:
```python
a = 'Hello'
b = 'World'
print(a + b)
# Output:
# HelloWorld
````

# Global Variables


What are global variables?

- Global variables are variables that are defined outside of any function or class and can be accessed from anywhere in the code, including inside functions and classes.
  They have a global scope, meaning they can be used throughout the entire program. However, it's generally recommended to use global variables sparingly, as they can make code harder to understand and maintain.


`````python
x = 'awesome'


def myfunc():
    print("Python is " + x)


myfunc()
`````

`````python
x = 'value1'
print('global value: ' + x)


def myfunc():
    x = 'value2'
    print("local value: " + x)


myfunc()
`````

# Global Keyword


What is the global keyword?

- The `global` keyword is used inside a function to indicate that a variable is global, meaning it refers to the variable defined outside the function.


`````python
def myfunc():
    global x
    x = "value1"
    print("value inside function:", x)


myfunc()
print("value outside of function:", x)
`````

When else do you use the global keyword?

- if you want to change a global variable INSIDE a function


`````python
x = "value1"
print("before:", x)


def myfunc():
    global x
    x = 'value2'


myfunc()

print("after:", x)
`````
