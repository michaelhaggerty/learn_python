# Data Types


What are data types?

- classifications that tell a programming language how to interpret, store, and interact with a specific piece of value - what operations can be performed and how it behaves in memory


What are the Built-in data types?

- Text Type: str
- Numeric Types: int, float, complex
- Sequence Types: list, tuple, range
- Mapping Type: dict
- Set Types: set, frozenset
- Boolean Type: bool
- Binary Type: bytes, bytearray, memoryview
- None Type: NoneType


How to get the type of any object?

- use the type() function


when is the data type set?

- when you assign a value to a variable


```python
# Example of type() function

x = 5
type(x)
```

create variables using literals:


```python
# str
x = "Hello, World!"

# int
x = 20

# float
x = 20.5

# complex
x = 1 + 2j

# list
x = ["a", "b", "c"]

# tuple
x = ("a", "b", "c")

# range
x = range(6)

# dict
x = {"name": "john", "age": 31}

# set
x = {"a", "b", "c"}

# frozen set
x = frozenset({"a", "b", "c"})

# bool
x = True

# bytes
x = b"Hello"

# bytearray
x = bytearray(5)

# memoryview
x = memoryview(bytes(5))

# NoneType
x = None
```

```python
x = bool(100)
x
```

```python
y = bool(0)
y
```

What are literals?

- when you assign a value to a variable, the value is called a literal. A literal is a notation for representing a fixed value in source code. For example, in the assignment `x = 5`, the number `5` is an integer literal. Similarly, in `y = "Hello"`, the string `"Hello"` is a string literal.

What are constructors?

- constructors are functions that allow you to create objects of a specific data type.

What is the difference between a constructor and casting?

- A constructor is a function that creates an object of a specific data type, while casting is the process of converting a value from one data type to another. In Python, constructors are used to create new instances of data types, while casting is used to change the type of an existing value.
- When you cast in Python, you are simply asking a class constructor to build a new version of your data in its specific format.


Creating variables using constructors:


```python
# str
x = str("Hello World")

# int
x = int(20)

# float
x = float(20.5)

# complex
x = complex(1j)

# list
x = list(("apple", "banana", "cherry"))

# tuple
x = tuple(("apple", "banana", "cherry"))

# range
x = range(6)

# dict
x = dict(name="John", age=36)

# set
x = set(("apple", "banana", "cherry"))

# frozenset
x = frozenset(("apple", "banana", "cherry"))

# bool
x = bool(5)

# bytes
x = bytes(5)

# bytearray
x = bytearray(5)

# memoryview
x = memoryview(bytes(5))
```
