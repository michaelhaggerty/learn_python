# Data Types


What are data types?

- classifications that tell a programming language how to interpret, store, and interact with a specific piece of value - what operations can be performed and how it behaves in memory


What are the Built-in data types?

- Text Type: str
- Numeric Types: int, float, complex
- Sequence Types: list, tuple, range
- Mapping Type: dict
- Set Types: set, frozenset
- Boolean Type: bool
- Binary Type: bytes, bytearray, memoryview

What are literals?

- when you assign a value to a variable, the value is called a literal. A literal is a notation for representing a fixed value in source code. For example, in the assignment `x = 5`, the number `5` is an integer literal. Similarly, in `y = "Hello"`, the string `"Hello"` is a string literal.

What are constructors?

- constructors are functions that allow you to create objects of a specific data type.

What is the difference between a constructor and casting?

- A constructor is a function that creates an object of a specific data type, while casting is the process of converting a value from one data type to another. In Python, constructors are used to create new instances of data types, while casting is used to change the type of an existing value.
- When you cast in Python, you are simply asking a class constructor to build a new version of your data in its specific format.


Creating variables using constructors:


```python
# str
x = str("Hello World")

# int
x = int(20)

# float
x = float(20.5)

# complex
x = complex(1j)

# list - outputs mutable array ['a', 'b', 'c']
x = list(("a", "b", "c"))

# tuple - outputs immutable array ('a', 'b', 'c')
x = tuple(("a", "b", "c"))

# range
x = range(6)

# dict
x = dict(name="John", age=36)

# set
x = set(("apple", "banana", "cherry"))

# frozenset
x = frozenset(("apple", "banana", "cherry"))

# bool
x = bool(5)

# bytes
x = bytes(5)

# bytearray
x = bytearray(5)

# memoryview
x = memoryview(bytes(5))
```

What's the deal with inputs of list() and tuple(), what kind of inputs can they take?
- they can both take iterable sequences

What are iterables? Sequences?
- An iterable in Python is any data object that contains a collection of items and allows you to loop over them one by one. 
- A sequence is a specific type of iterable where those items are stored in a strict mathematical order, allowing you to access them by their position or index (e.g., item 0, item 1).
- When we say a constructor like list() accepts an "iterable," it means it can take almost any structured collection of data and break it down item by item to build the new list.

What are some common iterables in Python?
- Lists ([]): Ordered, changeable sequences.
- Tuples (()): Ordered, unchangeable sequences.
- Strings (""): Ordered sequences of individual characters.
- Sets ({}): Unordered collections of unique items.
- Dictionaries ({key: value}): Collections of key-value pairs.

What happens if you pass a string to a list constructor?
- Because constructors accept any iterable, you are not limited to just passing brackets or parentheses. If you pass a string into a list constructor—like list("apple")—Python will iterate through that sequence one character at a time, building a new list out of the pieces: ['a', 'p', 'p', 'l', 'e'].

# Numbers

What are the three numeric types in Python?
* int = integer = whole number, positive or negative, without decimals, unlimited length
* float = "floating point number" = number, positive or negative, containing one or more decimals.
    * can also be scientific numbers with an "e" to indicate the power of 10.
- complex = Complex numbers --> written with a "j" as the imaginary part

```python
# examples
a = 1
b = 343434323343435467
c = -34356
print(type(a))
print(type(b))
print(type(c))

d = 1.5
e = 1.0
f = -45.78
g = 35e3
h = -87E100
print(type(d))
print(type(e))
print(type(f))
print(type(g))
print(type(h))

i = 1j
j = 3 + 5j
print(type(i))
print(type(j))
```

# Casting


What is casting?
- the process of manually converting a piece of data from one specific type to another

Why is this done in python?
- data types dictate how a program behaves, casting allows you to force data to act differently when needed, such as changing a text string of a number into an actual integer so you can perform math on it. 

How is this done?
- passing the original value into the constructor function of the target data type

What are the constructor functions?
- int()
    - int from float by chopping off decimal
    - int from string IF it contains valid whole number
- float()
    - constructs floating point from integer or valid string
- str()
    - constructs string from variety of data types

```python
# examples

x = int(2.5)
print(x)
print(type(x))

y = float("5")
print(y)
print(type(y))

z = str(10)
print(z)
print(type(z))
```

```python
a = int(1)   # a will be 1
b = int(2.8) # b will be 2
c = int("3") # c will be 3

d = float(1)     # d will be 1.0
e = float(2.8)   # e will be 2.8
f = float("3")   # f will be 3.0
g = float("4.2") # g will be 4.2

h = str("s1") # h will be 's1'
i = str(2)    # i will be '2'
j = str(3.0)  # j will be '3.0'
```
