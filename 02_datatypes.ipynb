# Slicing Strings

What is slicing?
- a technique used to extract a specific subset of items from a sequence (ex: string, list, or tuple)
- Instead of pulling out just one item at a time with a single index, slicing allows you to grab a continuous range or a patterned sequence of elements all at once.

How do you perform slicing?
- using square brackets and colons, with the pattern: [start:stop:step]
    - Start: The index where the slice begins (inclusive). If left blank, it defaults to the very beginning (index 0).

    - Stop: The index where the slice ends (exclusive). The slice extracts items right up to, but not including, this position. If left blank, it defaults to the very end of the sequence.

    - Step: The jump size between items. If left blank, it defaults to 1 (grabbing every consecutive item).

How do you slice from the start?
- By leaving out the start index, the range will start at the first character:

````python
b = "Hello, World!"
print(b[:5])
````

What happens if you leave out the end index?
- By leaving out the end index, the range will go to the end:

````python
b = "Hello, World!"
print(b[7:])
````

What is negative indexing?

````python
word = "Python"

# Accessing a single item from the end
print(word[-1]) 
# Output: n 

# Slicing with negative indexes
print(word[-4:-1]) 
# Output: tho 
# (Starts 4 spaces from the end 't', stops right before the last letter 'n')

print(word[2:5])
````

````python
text = "Hello, World!"

# Basic slice from index 0 to 5 (grabs indices 0, 1, 2, 3, 4)
print(text[0:5]) 
# Output: Hello

# Omitting the start index (starts from 0)
print(text[:5]) 
# Output: Hello

# Omitting the stop index (goes to the end)
print(text[7:]) 
# Output: World!

# Using the step parameter to skip every other character
print(text[0:5:2]) 
# Output: Hlo

# Negative indexing to slice from the end of the string
print(text[-6:-1]) 
# Output: World
````

What is the difference between inclusive and exclusive?
- The difference comes down to whether the specific boundary number is kept in your final result or left out.

    - Inclusive: The boundary point is included in the final result. If a range is 1 to 5 inclusive, it contains 1, 2, 3, 4, and 5.

    - Exclusive: The boundary point is excluded from the final result; the sequence stops exactly one step before it. If a range is 1 to 5 exclusive, it contains 1, 2, 3, and 4.

Is the start always inclusive?
- Yes, in standard Python, the start boundary is virtually always inclusive.

- Whether you are slicing a list, indexing a string, or generating numbers with the range() function, Python's built-in tools are designed to capture the exact starting point you specify. If you tell Python to start at index 2, the data at index 2 will always be the very first item in your result.

- This is a core design choice that pairs perfectly with Python's zero-based indexing and the "exclusive stop" rule to make programming logic much cleaner. Because the start is inclusive and the stop is exclusive, you can instantly calculate exactly how many items a slice will contain just by subtracting the start from the stop. For example, a slice from index 2 to index 5 ([2:5]) will always return exactly 3 items (5 - 2 = 3).

- If Python allowed the start boundary to be exclusive, doing the math to figure out the length of a slice or splitting data into clean, non-overlapping chunks would become significantly more complicated.

Why the starting at zero and using an inclusive start with an exclusive stop?
- This concept is known in mathematics and computer science as a half-open interval, and it was famously championed by programming pioneer Edsger W. Dijkstra(DŸKE-struh) in 1982.
- Dijkstra argued that starting at zero and using an inclusive start with an exclusive stop is the most elegant way to handle data sequences because it prevents "off-by-one" errors.

What benefits does Dijkstra's half-open interval have?
- This design provides two major practical benefits for programmers:

1. Instant length calculation: As you noted, the number of elements is always exactly stop - start. If both boundaries were inclusive, you would constantly have to add 1 to your mental math (stop - start + 1).

2. Clean sequence splitting: When you divide a dataset into consecutive chunks, the exclusive stop of the first chunk perfectly matches the inclusive start of the second chunk. Slicing [0:3] and then [3:6] splits the first six items cleanly. No items are skipped, no items are duplicated, and you never have to write messy code like [0:2] and [3:5].

Python's creator, Guido van Rossum, explicitly adopted this logic to make the language's syntax as intuitive and error-resistant as possible.

````python
word = "Python"

# The slice [1:4] starts at index 1 (inclusive) and stops at index 4 (exclusive).
# It grabs indices 1, 2, and 3. It does NOT grab index 4 ('o').
print(word[1:4])
# Output: yth
````

# Modifying Strings

What can we do with strings?

What bigger ideas does this connect to?
- Because strings in Python are perfectly immutable, these "modification" methods never actually change the original string in memory. Instead, methods like .upper(), .strip(), or .replace() read the original data, apply the changes, and output a completely new string object.

If you run a = "Hello" and then simply call a.upper(), the original variable a remains exactly "Hello". To actually save the modification, you must reassign the variable to the new object: a = a.upper().

Beyond immutability, this tutorial connects to two other major programming concepts:

1. Object-Oriented Programming (OOP): In Python, every piece of data is an "object." The dot syntax (like a.split()) is how you call a method—a specialized function permanently attached to the string class. Python already knows how to split or uppercase the text because the string object carries those tools with it.

2. Data Sanitization: In real-world software, inputs are messy. Users accidentally add spaces before typing their passwords, or they capitalize emails differently. Methods like .strip() (removing accidental spaces) and .lower() (standardizing capitalization) are the primary tools developers use to clean and normalize raw data before saving it to a database or analyzing it.

````python
 # return string in upper case
a = "Hello, World!"
print(a.upper()) 
````

````python
# return string in lower case
a = "Hello, World!"
print(a.lower())
````

# Python Escape Characters

### 1. What are Syntax Protectors?
Sequences allow you to type reserved syntax characters as literal text without accidentally ending a string or breaking your code.

| Code | Description | Purpose | Example |
| :--- | :--- | :--- | :--- |
| `\'` | Single Quote | Prevents a quote from terminating the string | `print('It\'s raining.')` |
| `\\` | Backslash | Escapes the backslash itself to print it literally | `print("C:\\Users\\Admin")` |

### 2. Whitespace & Formatting?
These sequences inject structural spacing that cannot be easily typed on a single line.

| Code | Description | Purpose | Example |
| :--- | :--- | :--- | :--- |
| `\n` | New Line | Drops the text that follows onto a new line | `print("Row 1\nRow 2")` |
| `\t` | Tab | Inserts a standard horizontal tab indentation | `print("Name:\tAlice")` |

### 3. Legacy Terminal Controls?
Originally used for mechanical typewriters and printers, these are now primarily used to control how text updates dynamically in a command-line terminal.

| Code | Description | Purpose | Example |
| :--- | :--- | :--- | :--- |
| `\r` | Carriage Return | Returns cursor to start of line, overwriting previous text | `print("Hello\rWorld")` outputs `World` |
| `\b` | Backspace | Deletes the single character immediately before it | `print("Hello\b World")` outputs `Hell World` |
| `\f` | Form Feed | Originally ejected a printer page; rarely used today | `print("Page 1\fPage 2")` |

### 4. Numeric Encodings
These sequences allow you to generate characters by typing their exact mathematical value (ASCII/Unicode), which is useful for foreign characters or low-level byte data.

| Code | Description | Purpose | Example |
| :--- | :--- | :--- | :--- |
| `\ooo` | Octal Value | Generates a character via its base-8 numerical code | `print("\110")` outputs `H` |
| `\xhh` | Hex Value | Generates a character via its base-16 numerical code | `print("\x48")` outputs `H` |

````python
# Octal code for the letter 'H' is 110
print("\110") 
# Output: H

# Hex code for the letter 'H' is 48
print("\x48") 
# Output: H
````

````python
txt = "Hi Sam!"
x = "mSa"
y = "eJo"
mytable = str.maketrans(x, y)
print(txt.translate(mytable))
````

What does the code above do?

This code performs a one-to-one character swap. maketrans() creates an instruction manual (a translation table) for swapping specific letters, and translate() applies those instructions to your string.

Here is exactly how the logic breaks down:

1. Creating the Map (str.maketrans(x, y))
The maketrans method looks at the two strings x and y and pairs up their characters based on their EXACT POSITIONS. The first letter of x will be replaced by the first letter of y, the second by the second, and so on.

"m" is mapped to "e"

"S" is mapped to "J"

"a" is mapped to "o"

(Note: Because of this direct pairing, the x and y strings must always be the exact same length, otherwise Python will throw an error).

2. Applying the Map (txt.translate(mytable))
The translate() method takes that mapping table and scans through your original string ("Hi Sam!") character by character.

It sees "H", "i", and " " (space). These are not in the map, so it leaves them alone.

It sees "S" and changes it to "J".

It sees "a" and changes it to "o".

It sees "m" and changes it to "e".

It sees "!", which is not in the map, and leaves it alone.

The Result:
When you run print(txt.translate(mytable)), the output will be:Hi Joe!

# String Methods

````python
# Python Built-in String Methods

Because strings are immutable, these built-in methods never change the original string in memory. Instead, they read the data, perform the specified action, and return a brand-new string (or a boolean/list/tuple depending on the method). 

### 1. Case Conversion
These methods standardize or alter the capitalization of your text.

| Method         | Description                                                                        | Example                                 |
| :------------- | :--------------------------------------------------------------------------------- | :-------------------------------------- |
| `capitalize()` | Converts the first character to upper case and the rest to lower case.             | `"hello".capitalize()` ➔ `'Hello'`      |
| `casefold()`   | Converts string into lower case (more aggressive than `lower()`).                  | `"ß".casefold()` ➔ `'ss'`               |
| `lower()`      | Converts a string into standard lower case.                                        | `"HELLO".lower()` ➔ `'hello'`           |
| `upper()`      | Converts a string into upper case.                                                 | `"hello".upper()` ➔ `'HELLO'`           |
| `swapcase()`   | Swaps cases, lower case becomes upper case and vice versa.                         | `"Hi There".swapcase()` ➔ `'hI tHERE'`  |
| `title()`      | Converts the first character of each word to upper case.                           | `"hello world".title()` ➔ `'Hello World'`|

### 2. Searching & Replacing
These methods help you find specific characters, count them, or swap them out.

| Method         | Description                                                                        | Example                                 |
| :------------- | :--------------------------------------------------------------------------------- | :-------------------------------------- |
| `count()`      | Returns the number of times a specified value occurs in a string.                  | `"apple".count("p")` ➔ `2`              |
| `find()`       | Searches the string for a value and returns the index. Returns `-1` if not found.  | `"hello".find("l")` ➔ `2`               |
| `rfind()`      | Searches from the right, returning the *last* position of where a value was found. | `"hello".rfind("l")` ➔ `3`              |
| `index()`      | Same as `find()`, but raises a `ValueError` if the item is not found.              | `"hello".index("e")` ➔ `1`              |
| `rindex()`     | Same as `rfind()`, but raises a `ValueError` if the item is not found.             | `"hello".rindex("l")` ➔ `3`             |
| `startswith()` | Returns `True` if the string starts with the specified value.                      | `"Hello".startswith("H")` ➔ `True`      |
| `endswith()`   | Returns `True` if the string ends with the specified value.                        | `"Hello".endswith("x")` ➔ `False`       |
| `replace()`    | Returns a string where a specified value is replaced with another.                 | `"cat".replace("c", "b")` ➔ `'bat'`     |

### 3. Splitting, Partitioning & Joining
These methods break strings apart into arrays/tuples or glue them together.

| Method         | Description                                                                        | Example                                 |
| :------------- | :--------------------------------------------------------------------------------- | :-------------------------------------- |
| `split()`      | Splits the string at a specified separator, and returns a list.                    | `"a,b,c".split(",")` ➔ `['a', 'b', 'c']`|
| `rsplit()`     | Splits the string starting from the right.                                         | `"a,b,c".rsplit(",", 1)` ➔ `['a,b', 'c']`|
| `splitlines()` | Splits the string at line breaks (`\n`) and returns a list.                        | `"a\nb".splitlines()` ➔ `['a', 'b']`    |
| `partition()`  | Splits string at the first occurrence of a separator into a 3-part tuple.          | `"1+2".partition("+")` ➔ `('1', '+', '2')`|
| `rpartition()` | Same as `partition()`, but splits at the *last* occurrence of the separator.       | `"1+2+3".rpartition("+")` ➔ `('1+2', '+', '3')`|
| `join()`       | Joins the elements of an iterable to the end of the string used as the separator.  | `"-".join(["a", "b"])` ➔ `'a-b'`        |

### 4. Whitespace, Formatting & Alignment
These methods trim unwanted spaces or pad your text to fit specific lengths.

| Method         | Description                                                                        | Example                                 |
| :------------- | :--------------------------------------------------------------------------------- | :-------------------------------------- |
| `strip()`      | Returns a trimmed version of the string (removes whitespace from both ends).       | `" hi ".strip()` ➔ `'hi'`               |
| `lstrip()`     | Returns a left-trim version of the string.                                         | `" hi ".lstrip()` ➔ `'hi '`             |
| `rstrip()`     | Returns a right-trim version of the string.                                        | `" hi ".rstrip()` ➔ `' hi'`             |
| `center()`     | Returns a centered string, padded with a specified character.                      | `"hi".center(6, "-")` ➔ `'--hi--'`       |
| `ljust()`      | Returns a left-justified version of the string, padded to the right.               | `"hi".ljust(5, "-")` ➔ `'hi---'`        |
| `rjust()`      | Returns a right-justified version of the string, padded to the left.               | `"hi".rjust(5, "-")` ➔ `'---hi'`        |
| `zfill()`      | Fills the string with a specified number of `0` values at the beginning.           | `"42".zfill(5)` ➔ `'00042'`             |
| `expandtabs()` | Sets the tab size of the string (replaces `\t` with spaces).                       | `"a\tb".expandtabs(4)` ➔ `'a   b'`      |
| `format()`     | Formats specified values and inserts them into string placeholders `{}`.           | `"Cost: ${}".format(5)` ➔ `'Cost: $5'`  |
| `format_map()` | Formats specified values using a dictionary mapping.                               | `"I'm {age}".format_map({'age': 31})` ➔ `'I'm 31'`|

### 5. Validation (Boolean `is...` Methods)
These methods interrogate the string and return `True` or `False` based on its contents.

| Method          | Description                                                                       | Example                                 |
| :-------------- | :-------------------------------------------------------------------------------- | :-------------------------------------- |
| `isalnum()`     | Returns `True` if all characters are alphanumeric (letters and numbers only).     | `"Data1".isalnum()` ➔ `True`            |
| `isalpha()`     | Returns `True` if all characters are in the alphabet (no numbers).                | `"Data".isalpha()` ➔ `True`             |
| `isascii()`     | Returns `True` if all characters are standard ASCII characters.                   | `"Hello".isascii()` ➔ `True`            |
| `isdecimal()`   | Returns `True` if all characters are decimals (0-9).                              | `"123".isdecimal()` ➔ `True`            |
| `isdigit()`     | Returns `True` if all characters are digits (includes subscripts like `²`).       | `"123²".isdigit()` ➔ `True`             |
| `isnumeric()`   | Returns `True` if all characters are numeric (includes fractions like `½`).       | `"½".isnumeric()` ➔ `True`              |
| `isidentifier()`| Returns `True` if the string is a valid Python variable name (identifier).        | `"my_var".isidentifier()` ➔ `True`      |
| `islower()`     | Returns `True` if all characters are lower case.                                  | `"hello".islower()` ➔ `True`            |
| `isupper()`     | Returns `True` if all characters are upper case.                                  | `"HELLO".isupper()` ➔ `True`            |
| `istitle()`     | Returns `True` if the string follows the rules of a title.                        | `"Hello World".istitle()` ➔ `True`      |
| `isspace()`     | Returns `True` if all characters in the string are whitespaces.                   | `"   ".isspace()` ➔ `True`              |
| `isprintable()` | Returns `True` if all characters are printable (excludes escape sequences).       | `"hi\n".isprintable()` ➔ `False`        |

### 6. Encoding & Translation
These methods handle byte-level conversions and complex character mapping.

| Method          | Description                                                                       | Example                                 |
| :-------------- | :-------------------------------------------------------------------------------- | :-------------------------------------- |
| `encode()`      | Returns a byte-encoded version of the string (defaults to UTF-8).                 | `"hi".encode()` ➔ `b'hi'`               |
| `maketrans()`   | Returns a translation table to be used in the `translate()` method.               | `str.maketrans("a", "b")` ➔ `{97: 98}`  |
| `translate()`   | Returns a translated string mapping characters using a translation table.           | `"cat".translate(table)` ➔ `'cbt'`      |
````

````python
txt = "Good night Sam!"
x = "mSa"
y = "eJo"
z = "odnght"
mytable = str.maketrans(x, y, z)
print(txt.translate(mytable))

````

Explain the code above:
When you pass a third string (z) into str.maketrans(x, y, z), it tells Python which characters to completely delete from the string.

The Swaps (x and y): Just like before, 'm' becomes 'e', 'S' becomes 'J', and 'a' becomes 'o'.

The Deletions (z): Every character listed in z ("o", "d", "n", "g", "h", "t") is targeted for removal.

Execution: Python scans "Good night Sam!". It swaps the letters for "Sam" into "Joe", deletes every instance of o, d, n, g, h, and t, and leaves spaces and punctuation alone.

The resulting output is:G i Joe!

````python
txt = "Good night Sam!"
mydict = {109: 101, 83: 74, 97: 111, 111: None, 100: None, 110: None, 103: None, 104: None, 116: None}
print(txt.translate(mydict))
````

Explain the code above:
This snippet proves that str.maketrans() isn't a magical black box—it is simply a convenient tool that generates a standard Python dictionary.

Computers don't actually read letters like 'm' or 'S'; they read their underlying numerical Unicode (ASCII) codes:

109 is the numeric code for 'm', and 101 is 'e' (so 109: 101 swaps m to e).

83 is 'S', and 74 is 'J'.

97 is 'a', and 111 is 'o'.

Setting a key to None (like 111: None for 'o') is Python's dictionary instruction to delete that character.

Passing a raw dictionary directly into .translate() produces the exact same output as using maketrans:G i Joe!
