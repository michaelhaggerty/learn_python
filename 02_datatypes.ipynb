# Data Types

What are data types?
- classifications that tell a programming language how to interpret, store, and interact with a specific piece of value - what operations can be performed and how it behaves in memory

What are the Built-in data types?
- Text Type: str
- Numeric Types: int, float, complex
- Sequence Types: list, tuple, range
- Mapping Type: dict
- Set Types: set, frozenset
- Boolean Type: bool
- Binary Type: bytes, bytearray, memoryview
- None Type: NoneType

How to get the type of any object?
- use the type() function

```python
# Example of type() function

x = 5
type(x)
```
