What's about division?
- Python has two division operators
    - / - Division (returns a float)
    - // - Floor division rounds DOWN to the nearest integer:

| Operator | Name | Example |
| :--- | :--- | :--- |
| `/` | Division | `x / y` |
| `//` | Floor Division | `x // y` |

So when // applied, positive numbers go to smaller number, and negative numbers go to more negative (smaller?)?

Yes, exactly! In mathematics, "smaller" always means lesser in value (further to the left on a number line).

Floor division (//) always rounds down toward negative infinity, which has a different effect depending on whether you start with a positive or negative number:

    Positive numbers: Rounding down moves the number toward zero, making it smaller.

        7 / 2 = 3.5 →
        → rounded down to 3 (3 is smaller than 3.5).

    Negative numbers: Rounding down moves the number away from zero, making it more negative (which also means smaller in value, because −4
    −4 is to the left of −3.5
    −3.5 on the number line).

        -7 / 2 = -3.5 →
        → rounded down to -4 (-4 is smaller/lesser than -3.5).

So your intuition is spot on: floor division always shifts the result to the next integer down on the number line, regardless of the sign.

````python
# With positive numbers (both give the same result)
print(7 // 2)   # Output: 3 (3.5 rounded down)

# With negative numbers (this is where they differ)
print(-7 // 2)  # Output: -4 (-3.5 rounded down toward negative infinity)

# Contrast with truncation (like converting a float to an int)
print(int(-3.5)) # Output: -3 (chopped off toward zero)
````

# Ternary Operator

What is the ternary operator?
- The ternary operator (more accurately called a conditional expression in Python) is a compact, one-line syntax used to evaluate a condition and select one of two values.

- Instead of writing a traditional multi-line if-else block just to assign a value to a variable, the ternary operator condenses the logic into a single line.

So what is a ternary operator essentially?
- "ternary operator" (because it takes three operands: the true value, the condition, and the false value), Python's official documentation and PEP 308 formally define it as a conditional expression.

Key Characteristics

    Short-Circuit Evaluation: Python evaluates the condition first. If it is True, it computes and returns the value_if_true and completely ignores the else branch. If it is False, it evaluates the else branch.

    Inline Return / Assignment: Because it evaluates to a value rather than acting as a control flow statement, you can use it anywhere an expression is expected—such as inside a print() function, a list comprehension, or a function return statement:
    Python

    print("Even" if num % 2 == 0 else "Odd")

    Readability Warning: While powerful for simple assignments, chaining multiple ternary operators together (e.g., A if cond1 else B if cond2 else C) severely degrades code readability and should be avoided in favor of standard if-elif-else blocks.

Give two example as normal conditional and ternary

````python
# normal
num = 6

if num > 5:
    x = "WEEKEND!"
else:
    x = "Workday"

print(x)
````

````python
# ternary
num = 6

x = 'Weekend' if num > 5 else 'Workday'

print(x)
````

Where else can a ternary operator be used?
- The ternary operator can be used instead of elif in longer if statements:

````python
num = 3

x = "Fri" if num == 5 else "Sat" if num == 6 else "Sun" if num == 7 else "weekday"

print(x)
````

# Comparison Operators

What are comparison operators?
- operators that are used to compare two values

What do they return?
- return True or False based on the comparison

Can you combine them?
- Yes, Python allows you to chain comparison operators

What are the comparison operators?

| Operator | Name | Example |
| :--- | :--- | :--- |
| `==` | Equal | `x == y` |
| `!=` | Not equal | `x != y` |
| `>` | Greater than | `x > y` |
| `<` | Less than | `x < y` |
| `>=` | Greater than or equal to | `x >= y` |
| `<=` | Less than or equal to | `x <= y` |

````python
x = 5
y = 3

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)

# combined
print(1 < x < 10)

print(1 < x and x < 10)
````

# Logical Operators

What are logical operators?
- keywords (and, or, not) used to combine conditional statements and evaluate boolean logic. Instead of evaluating numbers or strings directly like arithmetic operators, they evaluate truth values (True or False) to control decision-making in your code.

| Operator | Description | Example |
| :--- | :--- | :--- |
| `and` | Returns `True` if both statements are true | `x < 5 and x < 10` |
| `or` | Returns `True` if one of the statements is true | `x < 5 or x < 4` |
| `not` | Reverse the result, returns `False` if the result is true | `not(x < 5 and x < 10)` |

# Identity Operators

What are identity operators?
- operators are used to compare the objects, not if they are equal, but if they are actually the same object, with the same memory location:

| Operator | Description | Example |
| :--- | :--- | :--- |
| `is` | Returns `True` if both variables are the same object | `x is y` |
| `is not` | Returns `True` if both variables are not the same object | `x is not y` |

- The is operator returns True if both variables point to the same object:
- The is not operator returns True if both variables do not point to the same object

What's the difference between is and == ?
| Operator | Description |
| :--- | :--- |
| `is` | Checks if both variables point to the same object in memory |
| `==` | Checks if the values of both variables are equal |

````python
# is vs ==

x = [1, 2, 3]
y = [1, 2, 3]

print(x == y)
print(x is y)
````

### Q&A: Python Object Identity (`is`) vs. Value Equality (`==`) and Memory Management

#### **Q1: What is the result of `[1, 2] is [1, 2]`, and why?**
**A:** The result is **`False`**. 
* The `is` operator checks for **object identity**—whether both sides point to the exact same object at the exact same memory address (`id(a) == id(b)`).
* Because lists are **mutable** (changeable), Python creates a brand-new, independent list object in memory every time you write a list literal `[...]`. Even though both lists hold the same values, they reside at two different memory locations.

---

#### **Q2: How does `==` differ from `is` when comparing `[1, 2]` and `[1, 2]`?**
**A:** `[1, 2] == [1, 2]` evaluates to **`True`**.
* The `==` operator checks for **value equality** (whether the contents inside the objects are identical), whereas `is` checks whether they are the exact same object in memory.

---

#### **Q3: How can you verify object identity and memory addresses using variables in Python?**
**A:** You can inspect memory locations using the built-in `id()` function:
```python
a = [1, 2]
b = [1, 2]

print(a == b)        # True (identical contents)
print(a is b)        # False (different memory addresses)
print(id(a), id(b))  # Outputs two different memory IDs

# Assigning c to reference the exact same object as a:
c = a
print(a is c)        # True (both point to the same memory address)
