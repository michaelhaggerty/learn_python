What's about division?
- Python has two division operators
    - / - Division (returns a float)
    - // - Floor division rounds DOWN to the nearest integer:

| Operator | Name | Example |
| :--- | :--- | :--- |
| `/` | Division | `x / y` |
| `//` | Floor Division | `x // y` |

So when // applied, positive numbers go to smaller number, and negative numbers go to more negative (smaller?)?

Yes, exactly! In mathematics, "smaller" always means lesser in value (further to the left on a number line).

Floor division (//) always rounds down toward negative infinity, which has a different effect depending on whether you start with a positive or negative number:

    Positive numbers: Rounding down moves the number toward zero, making it smaller.

        7 / 2 = 3.5 →
        → rounded down to 3 (3 is smaller than 3.5).

    Negative numbers: Rounding down moves the number away from zero, making it more negative (which also means smaller in value, because −4
    −4 is to the left of −3.5
    −3.5 on the number line).

        -7 / 2 = -3.5 →
        → rounded down to -4 (-4 is smaller/lesser than -3.5).

So your intuition is spot on: floor division always shifts the result to the next integer down on the number line, regardless of the sign.

```python
# With positive numbers (both give the same result)
print(7 // 2)   # Output: 3 (3.5 rounded down)

# With negative numbers (this is where they differ)
print(-7 // 2)  # Output: -4 (-3.5 rounded down toward negative infinity)

# Contrast with truncation (like converting a float to an int)
print(int(-3.5)) # Output: -3 (chopped off toward zero)
```

```python
x = 15
y = 4

print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x % y)
print(x ** y)
print(x // y)
```
