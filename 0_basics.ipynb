# Python Comments


What are comments used for in Python

- to explain code
- to make code more readable
- to prevent execution of code when testing


```python
# comments start with "#" and python ignore content after number sign/pound sign/hashtag

print("I will run")

# example of prevent execution below
# print("I will not run")

# alternate location, stuff b4 runs
print("I will run, alternate comment location")
```

Can you do multi-line comments in Python?

- yes you can use multiple # signs in a row OR you can use triple quotes (""" or ''') to create a multi-line string that is not assigned to a variable, which effectively acts as a multi-line comment.

How does the multi-line with strings work?

- Python will ignore string literals that are not assigned to a variable,
